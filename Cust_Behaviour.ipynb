```python
import pandas as pd

df= pd.read_csv('customer_shopping_behavior.csv')
```

```python
df['Review Rating']= df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.columns= df.columns.str.lower()
df.columns=df.columns.str.replace(' ','_')
df=df.rename(columns={'purchase_amount_(usd)' : 'purchase_amount'})
```

```python
df.columns
```

```python
#create a new columns
labels=['young adults','Adults', 'Middle aged', 'Senior']
df['age_group']= pd.qcut(df['age'], q=4,labels=labels)
```

```python
df[['age','age_group']].head(10)
```

```python
 #purchase frequency days(column created)
frequency_mapping = {
'Fortnightly': 14,
'Weekly': 7,
'Monthly': 30,
'Quarterly': 90,
'Bi-Weekly': 14,
'Annually': 365,
'Every 3 Months': 90

}
df['frequency_purchase_days']= df['frequency_of_purchases'].map(frequency_mapping)

```

```python
df[['frequency_of_purchases','frequency_purchase_days']].head(10)
```
